```python
first = input("Enter first name: ").strip()
last = input("Enter last name: ").strip()
section = input("Enter section: ").strip()
subject = input("Enter favorite subject: ").strip()

full_name = (first + " " + last).title()

print("\nFormatted Output:")
print("Name:", full_name)
print("Name Code:", full_name.upper())
print("Section:", section.upper())
print("Favorite Subject:", subject.title())
print("Character Count:", len(full_name))
```
